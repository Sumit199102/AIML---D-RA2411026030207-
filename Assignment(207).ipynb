```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/loan_default.csv')
```

```python
print("Shape of data:", df.shape)
```

```python
display(df.info())
display(df.head(10))
```

The dataset contains 1,012 rows and 18 columns.

Numeric columns include age, income, loan_amount, credit_score, employment_years, debt_to_income, existing_loans, dependents, bank_balance, interest_rate, and previous_defaults.

Categorical columns include employment_type, marital_status, education, loan_purpose, and city_tier.

```python
display(df.describe())

categorical_cols = ['employment_type', 'marital_status', 'education', 'loan_purpose', 'city_tier']
for col in categorical_cols:
    print(f"\nValue counts for {col}:")
    print(df[col].value_counts())

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='loan_default')
plt.title('Distribution of Loan Defaults')
plt.show()
```

The target variable (loan_default) is imbalanced, with approximately 81.5% 'No' (repaid) and 18.5% 'Yes' (default) values.

This matters because a model could achieve high accuracy just by predicting the majority class, making evaluation metrics like recall and F1-score essential.

```python
print("Missing values per column:\n", df.isnull().sum())

if 'education' in df.columns:
    df['education'] = df['education'].fillna('Unknown')

initial_rows = len(df)
df = df.drop_duplicates()
print(f"Dropped {initial_rows - len(df)} duplicate rows.")

print("Negative ages:", (df['age'] < 0).sum())
print("Negative incomes:", (df['income'] < 0).sum())
```

Only the education column had missing values (61 missing entries).

We filled them with 'Unknown' to retain all customer records for training instead of dropping them.

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='loan_default', y='credit_score')
plt.title('Credit Score by Loan Default Status')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.barplot(data=df, x='city_tier', y='loan_default', ax=axes[0])
axes[0].set_title('Default Rate by City Tier')
axes[0].tick_params(axis='x', rotation=30)

sns.barplot(data=df, x='employment_type', y='loan_default', ax=axes[1])
axes[1].set_title('Default Rate by Employment Type')
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', annot_kws={'size': 8})
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

credit_score and debt_to_income appear to separate defaulters from non-defaulters best.

```python
df = df.drop(columns=['customer_id'])

df['loan_default'] = df['loan_default'].map({'Yes': 1, 'No': 0})

df = pd.get_dummies(df, drop_first=True)

X = df.drop('loan_default', axis=1)
y = df['loan_default']

print("Features shape:", X.shape)
```

Scikit-learn decision trees require numerical inputs because mathematical algorithms cannot compute splits on raw text strings directly.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)
```

We keep a test set aside to simulate unseen real-world data and evaluate how well the model generalizes without memorizing the training data.

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

train_acc = dt.score(X_train, y_train)
test_acc = dt.score(X_test, y_test)

print(f"Training Accuracy: {train_acc:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

There is a large gap between training accuracy (100%) and test accuracy, indicating that the untuned decision tree is overfitting.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report, roc_auc_score

y_pred = dt.predict(X_test)

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No Default', 'Default'])
disp.plot(cmap=plt.cm.Blues)
plt.show()

# Classification Report and ROC-AUC
print("Classification Report:\n", classification_report(y_test, y_pred))
y_prob = dt.predict_proba(X_test)[:, 1]
print(f"ROC-AUC Score: {roc_auc_score(y_test, y_prob):.4f}")
```

Approving a loan that defaults causes direct financial loss, making it worse than rejecting a good customer. This makes recall for defaults the most important metric.

```python
from sklearn.model_selection import GridSearchCV

for depth in [3, 5, 7, 10]:
    temp_dt = DecisionTreeClassifier(max_depth=depth, random_state=42)
    temp_dt.fit(X_train, y_train)
    print(f"Depth {depth} - Train Acc: {temp_dt.score(X_train, y_train):.4f} | Test Acc: {temp_dt.score(X_test, y_test):.4f}")

param_grid = {
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_leaf': [1, 5, 10, 20],
    'criterion': ['gini', 'entropy']
}

grid_search = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid, cv=5, scoring='f1')
grid_search.fit(X_train, y_train)

best_model = grid_search.best_estimator_
print("\nBest Parameters found:", grid_search.best_params_)
print(f"Tuned Model Test Accuracy: {best_model.score(X_test, y_test):.4f}")
```

A moderate depth (max_depth=5 or 7) gave the best test result. Tuning closed the gap between training and test scores, reducing overfitting.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import recall_score, f1_score, accuracy_score

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

lr = LogisticRegression(random_state=42)
lr.fit(X_train_scaled, y_train)

models = {
    'Logistic Regression': lr.predict(X_test_scaled),
    'Untuned Decision Tree': dt.predict(X_test),
    'Tuned Decision Tree': best_model.predict(X_test)
}

results = []
for name, preds in models.items():
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, preds),
        'Recall': recall_score(y_test, preds),
        'F1-Score': f1_score(y_test, preds),
        'ROC-AUC': roc_auc_score(y_test, lr.predict_proba(X_test_scaled)[:, 1] if 'Logistic' in name else (best_model.predict_proba(X_test)[:, 1] if 'Tuned' in name else dt.predict_proba(X_test)[:, 1]))
    })

comparison_df = pd.DataFrame(results)
display(comparison_df)
```

The tuned Decision Tree is recommended because it balances high interpretability with stable generalization on test data.

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(10, 5))
pd.Series(best_model.feature_importances_, index=X.columns).nlargest(10).plot(kind='barh')
plt.title('Top 10 Feature Importances')
plt.show()

```

```python
plt.figure(figsize=(15, 8))
plot_tree(best_model, feature_names=X.columns, class_names=['No', 'Yes'], filled=True, max_depth=3)
plt.show()
```

Top five features: credit_score, debt_to_income, income, previous_defaults, and loan_amount (matching the exploratory analysis).

Decision rule 1: If credit score is low and previous defaults > 0, predict default.

Decision rule 2: If credit score is high and debt-to-income is low, predict repayment.

The model effectively identifies high-risk applicants using core financial indicators. Its main limitation is class imbalance, which leads to false alarms. The model can be improved by collecting more historical data, handling class imbalance, or upgrading to a Random Forest.

```python
#EDA USING SALARY DATASET
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/salary_dataset (1) (1).csv")
df.head()
```

```python
#EXPERIENCE DISTRIBUTION
plt.figure(figsize=(8,5))
sns.histplot(df["Experience"],bins=10,kde=True)
plt.title("EXPERIENCE DISTRIBUTION")
plt.xlabel("EXPERIENCE")
plt.ylabel("EMPLOYEE")
plt.show()
```

```python
#Salary by JobRole
#Bi-variate analysis(including two variables)
plt.figure(figsize=(10,5))
sns.barplot(x="JobRole",y="Salary",data=df)
plt.title("SALARY BY JOBROLE")
plt.xlabel("JOBROLE")
plt.ylabel("SALARY")
plt.xticks(rotation=45)
plt.show()
```

```python
#Salary by Education
plt.figure(figsize=(8,5))
sns.barplot(x="Education",y="Salary",data=df)
plt.title("SALARY BY EDUCATION")
plt.xlabel("EDUCATION")
plt.ylabel("SALARY")
```

```python
#Salary by location
plt.figure(figsize=(10,5))
sns.barplot(x="Location",y="Salary",data=df)
plt.title("SALARY BY LOCATION")
plt.xlabel("LOCATION")
plt.ylabel("SALARY")
plt.show()
```

```python
#Experience Vs Salary
plt.figure(figsize=(10,5))
sns.scatterplot(x="Experience",y="Salary",data=df)
plt.title("EXPERIENCE VS SALARY")
plt.xlabel("EXPERIENCE")
plt.ylabel("SALARY")
plt.show()
```

```python
#Experience Vs Salary
plt.figure(figsize=(10,5))
sns.regplot(x="Experience",y="Salary",data=df)
plt.title("EXPERIENCE VS SALARY")
plt.xlabel("EXPERIENCE")
plt.ylabel("SALARY")
plt.show()
```

```python
#Salary distribution with education
plt.figure(figsize=(10,5))
sns.boxplot(x="Education",y="Salary",data=df)
plt.title("SALARY DISTRIBUTION WITH EDUCATION")
plt.xlabel("EDUCATION")
plt.ylabel("SALARY")
plt.show()
```
